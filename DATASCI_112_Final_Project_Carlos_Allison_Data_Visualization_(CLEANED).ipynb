### ALL VISUALIZATIONS USED ON THE POSTER

```python
import pandas as pd
df = pd.read_csv('combined_df.csv')
free_skate_songs = pd.read_csv('free_skate_songs.csv')
short_program_songs = pd.read_csv('short_program_songs.csv')
```

```python
df.set_index('Unnamed: 0')

df["sp_category"] = df["sp_category"].str.replace("_", "/").str.capitalize()
```

```python
df
```

```python
import plotly.express as px

color_map = {
    "Pop/contemporary": "#8b2fc9",
    "Warhorse": "#8ac926",
    "Movie/musical": "#1982c4"
}
```

```python
fig = px.histogram(
    df,
    x='sp_category',
    color='sp_category',
    color_discrete_map=color_map,
    title='Song Categories Amongst All Short Programs'
)

fig.update_xaxes(title_text="Music Category")
fig.update_yaxes(title_text="Number of Programs")
fig.update_layout(legend_title_text="Legend",
                  title=dict(
                  font=dict(size=25)))
fig.show()
```

```python
podium = df[df['SP Rank'] <= 3]

fig = px.histogram(podium, x='sp_category',
                   color='sp_category',
                   color_discrete_map=color_map,
                   title='Song Categories Amongst Top 3 Short Program Finishers')


fig.update_xaxes(title_text="Music Category")
fig.update_yaxes(title_text="Number of Programs")
fig.update_layout(legend_title_text="Legend",
                  title=dict(
                  font=dict(size=25)))
fig.show()
```

```python
top3_df = df[df["SP Rank"].isin([1, 2, 3])]

prop_df = (
    top3_df.groupby(["SP Rank", "sp_category"])
    .size()
    .reset_index(name="count")
)

prop_df["proportion"] = prop_df.groupby("SP Rank")["count"].transform(lambda x: x / x.sum())

fig = px.bar(
    prop_df,
    x="SP Rank",
    y="proportion",
    color="sp_category",
    barmode="stack",
    color_discrete_map=color_map,
    category_orders={
        "sp_category": ["Pop/contemporary", "Warhorse", "Movie/musical"],
        "SP Rank": [1, 2, 3],
    },
    title="Pop/Contemporary Music Dominates Podium",
    labels={
        "SP Rank": "Short Program Placement",
        "proportion": "Proportion of Songs",
        "sp_category": "Legend",
    },
)

fig.update_layout(
    plot_bgcolor="white",
    paper_bgcolor="white",
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
    xaxis=dict(tickvals=[1, 2, 3], ticktext=["1st", "2nd", "3rd"]),
    title=dict(font=dict(size=25))
)

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor="#eeeeee", tickformat=".0%")

fig.update_layout(
    xaxis=dict(tickmode='array', tickvals=[1, 2, 3], ticktext=["1st", "2nd", "3rd"]),
    bargap=0.3,        # increases space between groups → thinner bars
    bargroupgap=0.01   # small gap between stacked segments (optional)
)

fig.show()
```

```python
fig = px.box(df, x='sp_category', y='SP Score',
             color='sp_category',
             color_discrete_map=color_map,
             facet_col='Gender',
             points='all',
             title='Short Program Total Score by Gender')


fig.update_xaxes(title_text="Music Category")
fig.update_yaxes(title_text="SP Score")
fig.update_layout(legend_title_text="Legend",
                  title=dict(
                  font=dict(size=25)))
fig.show()
```

```python
df["skill_bin"], bins = pd.qcut(df["SP TES"], q=3, labels=["Low TES", "Mid TES", "High TES"], retbins=True)

median_df = (
    df.groupby(["skill_bin", "sp_category"])["SP PCS"]
    .median()
    .reset_index()
    .rename(columns={"SP PCS": "Median PCS"})
)

fig = px.bar(
    median_df,
    x="skill_bin",
    y="Median PCS",
    color="sp_category",
    color_discrete_map=color_map,
    barmode="group",
    category_orders={
        "sp_category": ["Pop/contemporary", "Warhorse", "Movie/musical"],
        "skill_bin": ["Low TES", "Mid TES", "High TES"],
    },
    title="Median PCS Grouped By TES",
    labels={
        "skill_bin": "Skill Level (TES)",
        "Median PCS": "Median Presentation Score (PCS)",
        "sp_category": "Music Category",
    }
)

fig.update_layout(
    plot_bgcolor="white",
    paper_bgcolor="white",
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
    bargap=0.25,
    bargroupgap=0.05,
)

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor="#eeeeee")

fig.update_xaxes(title_text="Technical Element Score Level")
fig.update_yaxes(title_text="Median Presentation Score (PCS)")
fig.update_layout(legend_title_text="Legend",
                  title=dict(
                  font=dict(size=25)))
fig.show()
```
