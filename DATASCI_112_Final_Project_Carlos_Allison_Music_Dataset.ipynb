# Utilizing apply function to map category in the dataframe.

```python
def categorize_song(song):
    for category, keywords in sp_mapping.items():
        for keyword in keywords:
            if keyword.lower() in song.lower():
                return category
    return 'Uncategorized'  # Default category for songs that don't match

# Apply the categorization function
df_sp_exploded['Category'] = df_sp_exploded['Song'].apply(categorize_song)

# Display the resulting DataFrame
print(df_sp_exploded)
```
